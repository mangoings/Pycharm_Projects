```python
import numpy as np

salary = [200, 150, 180, 220, 250]
```

```python
# 1. 1.5배 인상된 배열 구하기
new_salary = np.array(salary) * 1.5
print(f'인상된 금액 : {new_salary}')
```

```python
# 2. 평균 금액 구하기
mean_new_salary = new_salary.mean()
print(f'평균 금액 : {mean_new_salary}')
```

```python
# 3. 평균 이상인 원소
over_salary = new_salary[new_salary > mean_new_salary]
print(over_salary)
```

```python
# 4. 평균 이상인 원소의 개수
num_over_salary = new_salary.count()
print(num_over_salary)
```
