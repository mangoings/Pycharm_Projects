```python
# DataFrame()dmfh 생성한 데이터의 예
table_data1 = {'A': [1, 2, 3, 4, 5],
               'B': [10, 20, 30, 40, 50],
               'C': [100, 200, 300, 400, 500]}
df1 = pd.DataFrame(table_data1)
df1
```

```python
table_data2 = {'A': [6, 7, 8],
               'B': [60, 70, 80],
               'C': [600, 700, 800]}
df2 = pd.DataFrame(table_data2)
df2
```

```python
# 두 개의 DataFrame 데이터 df1과 df2는 길이가 같지 않음
# 길이가 같지 않더라도 연산이 가능. 결과값은 크기가 큰 데이터프레임의 크기와 같음.
df1 + df2
```

Series와 마찬가지로 DataFrame도 1) 연산할 수 있는 항목끼리만 연산하고 2) 그렇지 못한 항목은 NaN로 표시.
공간은 차지하되 데이터는 없는 상태

```python
# pandas에는 데이터의 통계 분석을 위한 다양한 메서드가 있어서 데이터의 총합, 평균, 표준편차등을 쉽게 구할 수 있음.
# 2021년부터 2016년까지 우리나라의 계절별 강수량(단위 mm)
table_data3 = {'봄': [256.5, 264.3, 215.9, 223.2, 312.8],
               '가을': [363.5, 231.2, 293.1, 247.7, 381.6],
               '여름': [770.6, 567.5, 599.8, 387.1, 446.2],
               '겨울': [139.3, 59.9, 76.9, 109.1, 108.1]}
colums_list = ['봄', '여름', '가을', '겨울']
index_list = ['2012', '2013', '2014', '2015', '2016']
df3 = pd.DataFrame(table_data3, columns=colums_list, index=index_list)
df3
```

pandas에서 제공하는 통계 매서드는 sum(), mean(), std(), var(), min(), max(), 각 원소의 누적 합을 구하는 cumsum(), 각 원소의 누적 곱을 구하는 cumprod() 등이 있음

```python
# 2012년에서 2016년에 걸쳐 계절별로 강수량의 평균(mean)과 표준편차(std)를 구함
df3.mean()
```

```python
df3.std()
```

```python
# 연도별로 평균 강수량과 표준편차를 구할 경우엔 연산의 방향 설정을 위해 axis 인자를 추가
# 인자 axis가 0이면 DataFrame의 values에서 열별로 연산을 수행하고
# 1이면 행별로 연산, axis 인자를 설정하지 않으면 기본값을 0으로 설정
df3.mean(axis=1)
```

```python
df3.std(axis=1)
```

```python
df3.T
```

```python
df3.T.mean()
```
