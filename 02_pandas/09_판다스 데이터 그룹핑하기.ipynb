```python
# 판다스 데이터 그룹핑하기
# 데이터 그룹핑은 데이터를 특정한 값에 기반해 묶는 기능으로 groupby()를 사용하여 통계량을 요약할 수 있음
# 통계량에 관계된 매서드에는 mean, std, var,max,min 등이 있음/
```

```python
import pandas as pd

df = pd.read_excel('./input/nation.xlsx')
df
```

```python
# 1. 한 열을 기준으로 그룹화하기
df.groupby('국적코드').sum()
```

```python
df.groupby('성별').sum()
```

```python
# groupby()를 사용하여 '국적코드' 열을 기준으로 그룹화하고, sum() 매서드를 적용
```

```python
# 2. 여러 열을 기준으로 그룹화하기
df.groupby(['국적코드', '성별']).sum()
```

```python
df.groupby(['성별', '국적코드']).sum()
```

```python
df.groupby(['성별', '국적코드']).mean()
```

```python
df.groupby('국적코드')[['입국객수', '증가수']].mean()
```
