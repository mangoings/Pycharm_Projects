```python
import pandas as pd
```

```python
df = pd.read_csv('ex_04.csv')
df.head()
```

```python
df.info()
```

```python
df.info(memory_usage='deep')
```

```python
# 2. 필요한 변수(컬럼)만으로 데이터프레임 생성
df1 = df[['상가업소번호', '상호명', '시군구명', '법정동명', '표준산업분류명']]
```

```python
df1.head()
```

```python
df1.info(memory_usage='deep')
```

```python
# 3. 커피전문점만 추출
df2 = df1[df1['표준산업분류명'] == ['커피전문점']]
```

```python
df2.head()
```

```python
df2.info(memory_usage='deep')
```

```python
# 4. 대구에 있는 커피 전문점(표준산업분류명 기준)을 구별로 몇개가 있는지 출력
df2.value_counts('시군구명')
```

```python
df2.value_counts('시군구명').sort_index(ascending=False)
```

```python
# 5. 대구에 있는 커피 전문점(표준산업분류명 기준)을 동(법정동 기준)별로 몇개가 있는지 출력
df2.value_counts('법정동명')
```

```python
df2.value_counts('법정동명').index
```

```python
df2.value_counts('법정동명').sort_index()
```
