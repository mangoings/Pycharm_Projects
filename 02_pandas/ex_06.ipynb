```python
import pandas as pd
```

```python
df = pd.read_excel('./input/vaccine.xlsx')
df.tail()
```

```python
df.info()
```

```python
# 필요없는 열을 삭제한 후 출력
# 삭제할 열 : day1sum, day1rate, day2sum, day2rate
df1 = df.drop(columns=['day1sum', 'day1rate', 'day2sum', 'day2rate'])
df1.head()
```

```python
#
df1 = df.drop(columns=['day1sum', 'day1rate', 'day2sum', 'day2rate'], axis=1)
df1.describe()
```

```python
# 기초통계량에 subject 컬럼을 출력하지 않으려면 문자열로 변환
df2 = df1.copy()
```

```python
df2['subject'] = df2['subject'].astype('str')
df2.head()
```

```python
df2.describe()
```

```python
# 새로울 열("월")을 생성한 후
df3 = df2.copy()
df3.head()
```

```python
df3.info()
```

```python
df3['월'] = df3['date'].str.slice(5, 7)
df3
```

```python
# 평균을 출력할 수 있는 열만 선택한 후 평균
df4 = df3[['day1num', 'day2num', '월']]
df4
```

```python
df4.shape
```

```python
df5 = df4.groupby('월').mean()
df5
```
