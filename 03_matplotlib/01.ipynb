```python
# matplotlib
# 맷플롯립은 파이썬에서 그래프를 그릴 때 가장 많이 사용하는 파이썬 라이브러리 중 하나.
# 맷플롯립을 활용하면 다양한 유형의 그래프를 간편하게 그릴 수 있음.
# 시각화할 때 데이터로 리스트나 딕셔너리 자료형도 가능하지만 데이터프레임을 이용하면
# 새로운 그래프에 적용하는 것이 더 쉽기 때문에 대부분 데이터프레임 자료형을 이용.

# matplotlib는 MATLAB (과학 및 공학 연산을 위한 소프트웨어)의 시각화 기능을 모델링해서 만들어짐.

# 공식 홈페이지 : https://matplotlib.org/
```

```python
import matplotlib.pyplot as plt
```

```python
# 1) 선 그래프
# 기본적인 선 그래프 그리기.
# 선 그래프는 그래프 중 가장 기본이 되는 그래프.
# 순서가 있는 숫자 데이터를 시각화하거나 시간에 따라 변하는 숫자 데이터를 시각화하는데 많이 사용.
# *기본 형식: plt.plot([x, ] y[, fmt])
# x와 y는 각각 x축과 y축 좌표의 값을 의미.
# x와 y는 시퀀스 길이가 같아야 함.
# x와 y 데이터 중에서 x는 생략 가능.
# x가 없다면
# x는 0부터 y의 개수만큼 1씩 증가하는 값으로 자동 할당.
# fmt는 format string으로 다양한 형식으로 그래프를 그릴 수 있는 옵션.
```

```python
# x축 데이터를 지정하지 않고 y축 데이터만 있는 경우
# 숫자로 이루어진 리스트 데이터 생성
data1 = [10, 14, 19, 20, 25]
plt.plot(data1) # x축 데이터가 생략
plt.show()
```

```python
# 1) x축 값을 지정안함. x의 범위가 0~y의 개수 -1
# 2) y축의 범위는 y의 최소값 ~ y의 최대값
```

```python
# x값과 y값이 모두 있는 데이터
import numpy as np

x = np.arange(-4.5, 5, 0.5) # ㅂㅐ열 x 생성. 범위: [-4.5, 5), 0.5씩 증가
y = 2 * x **2 # 수식을 이용해 배열 x에 대응하는 배열 y 생성
print([x, y])
```

```python
plt.plot(x, y)
plt.show()
```

```python
# 2. 여러 그래프 그리기
```

```python
# 여러개의 데이터를 효과적으로 시각화하는 방법
import numpy as np
x = np.arange(-4.5, 5, 0.5)
y1 = 2 * x ** 2
y2 = 5 * x + 30
y3 = 4 * x ** 2 + 10
```

```python
# 1) 하나의 그래프 창에 여러개의 그래프를 그리는 방법
```

```python
# 연속해서 plot()를 수행하면 하나의 그래프 창에 여러개의 그래프를 그릴 수 있음
plt.plot(x, y1)
plt.show()
plt.plot(x, y2)
plt.plot(x, y3)
plt.show()
```

```python
# 하나의 plot()에 표시하고자 하는 데이터를 모두 입력하는 방법으로도 하나의 그래프 창에 여러개의 그래프를 그릴 수 있음
plt.plot(x, y1, x, y2, x, y3)
plt.show()
```

```python
# 2) 여러개의 그래프 창에 만들어서 그래프를 그리는 방법

plt.plot(x, y1) # 자동으로 figure 명령어가 수행됨
plt.figure() # 새로운 그림판 생성
plt.plot(x, y2)
plt.show()
```

```python
# 그래프 창의 번호를 명시적으로 지정한 후 해당 창에 그래프를 그릴 수 있음
# plt.figure(n)

# 데이터 생성
x = np.arange(-5, 5, 0.1)
y1 = x ** 2 -2
y2 = 20 * np.cos(x) ** 2  # Numpy에서 cos()는 np.cos()으로 입력.
```

```python
plt.figure(1)
plt.plot(x, y1)

plt.figure(2)
plt.plot(x, y2)

plt.figure(1)
plt.plot(x, y2)

plt.figure(2)
plt.cla() # 2번 그래프 창에 그려진 모든 그래프를 지움
plt.plot(x, y1)

plt.show()
```

```python
# 3) 하나의 그래프 창을 하위 그래프 영역으로 나눈 후에 각 영역에 그래프를 그리는 방법
```

```python
# 하나의 그래프 창을 하위 그래프 영역으로 나누기 위해 subplot()을 이용.
# plt.subplot(m, n, p)
# m * n 행렬로 이뤄진 하위 그래프 중에서 p번 위치에 그래프가 그려지도록 지정.
# p는 왼쪽에서 오른쪽으로, 그리고 위에서 아래로 행렬의 위치를 지정.
```

```python
# 데이터 생성.
x = np.arange(0, 10, 0.1)
y1 = 0.3 * (x - 5) ** 2 + 1
y2 = 1.5 * x + 3
y3 = np.sin(x) ** 2  # Numpy에서 sin()은 np.sin()으로 입력.
y4 = 10 * np.exp(-x) + 1
```

```python
# 2x2 행렬로 이뤄진 하위 그래프에서 p에 따라 위치를 지정
plt.figure(figsize=(10, 10))

plt.subplot(2, 2, 1)
plt.plot(x, y1)

plt.subplot(2, 2, 2)
plt.plot(x, y2)

plt.subplot(2, 2, 3)
plt.plot(x, y3)

plt.subplot(2, 2, 4)
plt.plot(x, y4)

plt.show()
```

```python
# 3. 그래프의 출력범위 지정하기

# 그래프로 데이터를 분석할 때 전체 그래프 중 관심 영역만 확대해서 보고 싶을 때
# 출력되는 그래프의 x축과 y축의 좌표 범위를 지정해 전체 그래프 중 관심 영역만 그래프로 그릴 수 있음.

# plt.xlim(xmin, xmax)  # x축의 좌표 범위 지정 (xmin ~ xmax)
# plt.ylim(ymin, ymax)  # y축의 좌표 범위 지정 (ymin ~ ymax)
# xlim()과 ylim()은 독립적으로 동작하므로 둘 중 하나만 이용할 수도 있음.
```

```python
# 데이터 생성
x = np.linspace(-4, 4, 100)  # [-4, 4] 범위에서 100개의 값 생성.
y1 = x ** 3
y2 = 10 * x ** 2 -2
```

```python
plt.figure()
plt.plot(x, y1, x, y2)
plt.show()
```

```python
# 두 그래프가 만나는 부분이 잘 보이지 않아서 확대
plt.plot(x, y1, x, y2)
plt.xlim(-1, 1)
plt.ylim(-10, 10)
plt.show()
```
