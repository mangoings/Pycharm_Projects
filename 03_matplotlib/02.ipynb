```python
# plot()에서 사용할 수 있는 fmt 옵션을 활용해 그래프에서 선의 출력 형식을 지정하고
# 그래프에 축의 라벨, 제목, 격자, 문자열을 추가하는 방법
```

```python
import matplotlib
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1) 출력 형식 지정

# fmt 옵션을 이용하면 그래프의 컬러, 선의 스타일, 마커를 지정할 수 있음.

# fmt = '[color][line_style][maker]'

# color, line_style, maker는 각각 컬러, 선의 스타일, 마커 지정을 위한 약어(문자)를 나타냄.
# 이것은 하나씩만 지정할 수도 있고 조합해서 지정할 수도 있음.

# * 컬러 지정의 위한 약어.
# b	: 파란색(blue)
# g	: 녹색(green)
# r	: 빨간색(red)
# c	: 청녹색(cyan)
# m : 자홍색(magenta)
# y	: 노란색(yellow)
# k	: 검은색(black)
# w	: 흰색(white)

# * 선의 스타일 지정을 위한 약어
# -	  : 실선(solid line)
# --    :	파선(dashed line)
# :	  : 점선(dotted line)
# -.    :	파선 점선 혼합선(dash-dot line)

# * 마커 지정을 위한 약어
# o     :	원 모양 ●
# ^, v, <, >  :	삼각형 위쪽(▲), 아래쪽(▼), 왼쪽(◀), 오른쪽(▶)
# s	  : 사각형(square) ■
# p	  : 오각형(pentagon)
# h, H  : 육각형(hexagon)1, 육각형2
# *	  : 별 모양(star) ★
# +	  : 더하기(plus) +
# x, X  : ×, 채워진 ×
# D, d  : 다이아몬드(diamond, ◆), 얇은 다이아몬드
```

```python
x = np.arange(0, 5, 1)
y1 = x
y2 = x + 1
y3 = x + 2
y4 = x + 3
```

```python
plt.plot(x, y1, x, y2, x, y3, x, y4)
plt.show()
```

```python
# fmt 옵션을 지정하지 않아도 데이터에 따라 그래프의 선의 칼라가 다르게 그려짐

# fmt 옵션을 이용해 선 그래프에서 데이터 별로 선의 컬러를 지정하는 예

plt.plot(x, y1, 'm', x, y2, 'y', x, y3, 'k', x, y4, 'c')
plt.show()
```

```python
# 선의 스타일을 지정해 그래프를 그리는 예
plt.plot(x, y1, '-', x, y2, '--', x, y3, ':', x, y4, '-.')
plt.show()
```

```python
# (x, y) 데이터의 각 점을 선으로 연결하는 것이 아니라, 각 데이터가 표시다는 지점을 다양한 모양의 마커로 표시하는 예
plt.plot(x, y1, 'o', x, y2, '^', x, y3, 's', x, y4, 'd')
plt.show()
```

```python
# plot()의 fmt 옵션에서 컬러, 선의 스타일, 마커는 혼합해서 지정할 수 있음
plt.plot(x, y1, '>-r', x, y2, 's-g', x, y3, 'd:b', x, y4, '-.Xc')
plt.show()
```

```python
# 2. 라벻,제목,격자,범례,문자열 표시
# 그래프의 x축과 y축에 라벨,그래프 제목,격자,범례,문자열 표시
```

```python
x = np.arange(-4.5, 5, 0.5)
y = 2 * x ** 3
```

```python
plt.plot(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
# 그래프 제목은 title('문자열')로 추가할 수 있음
plt.plot(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Graph Title')
plt.show()
```

```python
# 그래프에 격자를 추가하려면 grid(True) 혹은 grid()를 이용
# 이미 격자가 있는 그래프에서 격자를 제거하고 싶으면 grid(False)를 이용
plt.plot(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Graph Title')
plt.grid(True)
plt.show()
```

```python
# *범례 legend: 하나의 그래프에 선이 여러개 있다면 선을 구별하기 어려울 때 사용
# 여러 선이 있는 그래프에서 legend([])을 이용해 범례를 이용
x = np.arange(0, 5, 1)
y1 = x
y2 = x + 1
y3 = x + 2
y4 = x + 3
plt.plot(x, y1, '>-r', x, y2, 's-g', x, y3, 'd:b', x, y4, '-.Xc')
plt.legend(['x1', 'x2', 'x3', 'x4'])
plt.show()
```

```python
#* loc 옵션으로 범례의 위치를 지정.
#loc='위치 문자열' 과 loc=위치코드를 입력하는 방식이 있음.

#* legend()에서 loc 옵션으로 범례의 위치 지정.
#범례 위치 : 위치 문자열 - 위치 코드
#최적 위치 자동 선정 (디폴트) : best - 0
#상단 우측 : upper right - 1
#상단 좌측 : upper left - 2
#하단 좌측 : lower left - 3
#하단 우측 : lower right - 4
#우측 : right - 5
#중앙 좌측 : center left - 6
#중앙 우측 : center right - 7
# 하단 중앙 : lower center - 8
# 상단 중앙 : upper center - 9
# 중앙 : center - 10
```

```python
plt.plot(x, y1, '>-r', x, y2, 's-g', x, y3, 'd:b', x, y4, '-.Xc')
plt.legend(['data1', 'data2', 'data3', 'data4'], loc='lower right')
plt.show()
```

```python
# 3. 한글 사용
# 그래프에서 한글을 표시하고 싶으면 matplotlib에서 사용하는 폰드를 한글 폰트로 지정해야함
```

```python
import matplotlib
print(matplotlib.rcParams['font.family'])
```

```python
# * 폰트를 변경하려면 다음과 같이 수행
matplotlib.rcParams['font.family'] = 'batang'
matplotlib.rcParams['axes.unicode_minus'] = False

# matplotlib.rcParams['axes.unicode_minus'] = False는 한글 폰트를 지정한 후에
# 그래프에서 마이너스(-) 폰트가 깨지는 문제를 해결해줌.
```

```python
plt.plot(x, y1, '>-r', x, y2, 's-g', x, y3, 'd:b', x, y4, '-.Xc')
plt.legend(['데이터1', '데이터2', '데이터3', '데이터4'], loc='best')
plt.xlabel('X 축')
plt.ylabel('Y 축')
plt.title('그래프 제목')
plt.grid(True)
plt.show()
```

```python
# 그래프 창에 좌표(x,y)를 지정해 문자열을 표시할 수도 있음
# plt.text(x, y, str)
plt.plot(x, y1, '>-r', x, y2, 's-g', x, y3, 'd:b', x, y4, '-.Xc')
plt.text(0, 6, '문자열 출력 1')
plt.text(0, 5, '문자열 출력 2')
plt.text(3, 1, '문자열 출력 3')
plt.text(3, 0, '문자열 출력 4')
plt.show()
```
