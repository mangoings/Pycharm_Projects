```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import pandas as pd
```

```python
matplotlib.rcParams['font.family'] = 'batang'
matplotlib.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.read_excel('./input/fine_dust.xlsx', index_col='area')
df
```

```python
df2019 = df[2019]
df2019.head()
```

```python
plt.figure(figsize=(15, 4))
plt.bar(df2019.index, df2019, color='b')
plt.title('2019 Fine Dust Line Graph')
plt.xlabel('area')
plt.ylabel('micrometer')
plt.grid(True)
plt.show()
```

```python
# 2) 그룹 세로 막대 그래프 그리기

# 서울, 경기, 인천, 부산 지역의 2016~2019년 데이터만 추출
df1 = df.loc['Seoul':'Busan', 2016:2019]
df1
```

```python
index = np.arange(len(df1))
bar_width = 0.2

plt.figure(figsize=(15, 5))
for i in range(2016, 2020):
    plt.bar(index, df1[i], width=bar_width, align='edge', label=i)
    index = index + bar_width
plt.xticks(index - 0.4, df1.index)
plt.ylim(35, 55)
plt.title('2016 ~ 2019 Fine Dust Line Graph')
plt.xlabel('area')
plt.ylabel('micrometer')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 3) 그룹 누적 가로 막대 그래프 만들기
# 2016 ~ 2019 지역별 미세먼지 그룹 막대 그래프를 누적하여 가로 방향으로 변경하여 표현.
# 누적 막대 그래프를 그리는 것은 기존 막대 그래프 위에 막대 그래프를 하나 더 그리는 형식.
# 이를 위해서는 index의 위치를 변경해야 함.
# 가로 방향으로 바꾸는 것은 bar() 대신 barh()를 사용하면 됨.
# 막대 그래프를 가로 방향으로 변경했으므로 x축 및 y축 이름, x축 및 y축 눈금도 가로 방향에 맞게 변경.
```

```python
index = np.arange(len(df1))

plt.figure(figsize=(15, 5))
for i in range(2016, 2020):
    plt.barh(index, df1[i], label=i)
plt.yticks(index, df1.index)
plt.title('2016 ~ 2019 Fine Dust Line Graph')
plt.ylabel('area')
plt.xlabel('micrometer')
plt.legend()
plt.grid(True)
plt.show()
```
