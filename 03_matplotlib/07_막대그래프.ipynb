```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

```python
matplotlib.rcParams['font.family'] = 'batang'
matplotlib.rcParams['axes.unicode_minus'] = False
```

```python
# * 막대 그래프 (bar 그래프) : 값을 막대의 높이로 나타냄. 여러 항목의 수량이 많고 적음을 한 눈에 알아 볼 수 있음.
# 여러 항목의 데이터를 서로 비교 할 때 주로 이용.
#
# plt.bar(x, height [width=width_f, color=colors, tick_label=tick_labels, align='center'(기본) 혹은 'edge', label=labels])
#
# height는 시각화하고자 하는 막대 그래프 데이터.
# x는 height와 길이가 일치하는 데이터로 'x축에 표시될 위치'를 지정.
# width 옵션으로 [0, 1] 사이의 실수를 지정해 막대의 폭을 조정.
# width 옵션을 입력하지 않으면 기본값인 0.8이 입력.
# color 옵션으로는 fmt 옵션의 컬러 지정 약어를 이용해 막대 그래프의 색을 지정.
# tick_label 옵션으로 문자열 혹은 문자 리스트를 입력해 막대 그래프 각각의 이름을 지정.
# align은 막대 그래프의 위치를 가운데로 할지 (center) 한쪽으로 치우치게 할지(edge)로 설정.
# 기본 옵션은 center.
# label에는 범례에 사용될 문자열을 지정.
```

```python
# 운동 시작전과 한달 후의 윗몸 일으키기 횟수.
# 회원 아이디 : 'm_01', 'm_02', 'm_03', 'm_04'
# 운동 시작 전 : 27, 35, 40, 33
# 운동 한 달 후 : 30, 38, 42, 37

member_ids = ['m_01', 'm_02', 'm_03', 'm_04'] # 회원 ID
before_ex = [27, 35, 40, 33] # 운동 시작 전
after_ex = [30, 38, 42, 37] # 운동 한달 후
```

```python
n_data = len(member_ids) # n_data = 4
index = np.arange(n_data) # 배열 생성 (0, 1, 2, 3)
index
```

```python
plt.bar(index, before_ex)
plt.show()
```

```python
# x축의 tick 라벨이 숫자로 되어 있는데 이 숫자는 막대 그래프 x축의 위치를 나타낼 뿐 의미가 없음
# tick_label 옵션을 이용하면 x축의 tick 라벨을 원하는 데이터로 지정
```

```python
plt.bar(index, after_ex, tick_label=member_ids)
plt.show()
```

```python
# 막대 그래프의 색깔을 변경하고 싶으면 color 옵션을 이용
colors = ['r', 'g', 'b', 'c']
plt.bar(index, before_ex, tick_label=member_ids, color=colors)
plt.show()
```

```python
# 막대 그래프의 넓이를 변경하려면 width 옵션 이용
plt.bar(index, after_ex, tick_label=member_ids, color=colors, width=0.6)
plt.show()
```

```python
# 가로 막대 그래프를 그리려면 barh()을 이용
# 사용방법은 bar()와 같으나 width 옵션을 이용할 수 없음
colors = ['r', 'g', 'b', 'c']
plt.barh(index, before_ex, tick_label=member_ids, color=colors)
plt.show()
```

```python
# 운동 시작 전과 한달 후의 데이터를 하나의 그래프 창에 막대 그래프로 그려서 비교
bar_width = 0.3
plt.bar(index - bar_width, before_ex, tick_label=member_ids, width=bar_width, align='edge', label = 'before', color='r')
plt.bar(index, after_ex, tick_label=member_ids, width=bar_width, align='edge', label = 'after', color='c')
plt.legend()
plt.xlabel('회원 ID')
plt.ylabel('윗몸 일으키기 횟수')
plt.title('운동 시작 전과 후의 근지구력(복근) 변화 비교')
plt.show()
```
