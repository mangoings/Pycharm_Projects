```python
# 서로 다른 두 연속성 변수 사이의 관계를 나타내며 연속성 변수의 상관성을 확인할 때 산점도 그래프를 사용.
# 예를 들어 나이와 소득에 대한 상호 관련성 파악 등에 유용하게 사용되는데, 두 변수 간의 상관관계를 개략적으로 파악할 수 있기 때문,
# 산점도에 표시되는 각 점들은 자료의 관측값을 의미하고, 각 점의 위치는 관측값이 가지는 x축, y축 변수의 값으로 결정
#
# 공공데이터포털 https://www.data.go.kr/ 에서 다운로드한 2020년 건강검진 일부 데이터 health_screenings_2020.xlsx를
# 이용하여 산점도 그래프를 표현
#
# https://www.data.go.kr/data/15007122/fileData.do
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./input/health_screenings_2020.csv')
df.head()
```

```python
df.info()
```

```python
df1 = df.iloc[:100000]
df1.head()
```

```python
df1.info()
```

```python
df_height = df1.height
df_weight = df1.weight
```

```python
df_weight.head()
```

```python
plt.figure(figsize = (10,5))
plt.scatter(df_height, df_weight)
plt.title('2020 Health Screening Data')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.show()
```

```python
# 2) 그룹 산점도 그래프 그리기

# HDL, LDL, cholesterol 데이터를 가져와 각각의 변수에 저장
HDL_data = df1['HDL']
LDL_data = df1['LDL']
chol_data = df1['cholesterol']

plt.figure(figsize = (10,6))
# scatter()에 x축 데이터와 y축 데이터를 지정하고 점 색상과 점 테두리 색상을 지정
plt.scatter(chol_data, LDL_data, color='r', edgecolor='w', label='Chol*LDL')
plt.scatter(HDL_data, chol_data, color='g', edgecolor='w', label='HDL*Chol')
plt.scatter(HDL_data, LDL_data, color='b', edgecolor='w', label='HDL*LDL')
plt.title('2020 Health Screening Data')
plt.xlim(-20, 500)
plt.ylim(-20, 500)
plt.legend()
plt.show()
```
