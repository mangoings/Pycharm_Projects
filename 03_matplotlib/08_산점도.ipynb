```python
# * 산점도 scatter 그래프는 두 개의 요소로 이뤄진 데이터 집합의 관계.
# (예를 들면, 키와 몸무게와의 관계, 기온과 아이스크림 판매량과의 관계, 공부 시간과 시험 점수와의 관계)를 시각화하는 데 유용.
#
# plt.scatter(x, y [,s=size_n, c=colors, marker='marker_string', alpha=alpha_f])
# x와 y는 각각 두 개의 데이터로 이루어진 데이터 쌍에서 x축과 y축 좌표의 값을 의미.
# 옵션인 s, c, marker, alpha를 이용해 각각 마커의 크기, 컬러, 모양, 투명도를 지정.
# 옵션을 지정하지 않으면 기본 값(s=36, c='b', marker='o', alpha=1)으로 지정.
#
# s는 원하는 크기의 값을 넣으면 되고
# c와 marker에는 fmt 옵션의 컬러와 마커 지정 약어를 선택해서 입력.
# s, c의 경우 하나만 입력하면 모든 마커에 동일하게 적용되고,
# 시퀀스를 입력하면 마커마다 다르게 적용.
# 투명도를 지정하는 alpha에는 [0, 1] 범위의 실수 입력.
# 0은 완전 투명이고, 1을 지정하면 완전 불투명.
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
height = [165, 177, 160, 180, 185, 155, 172] # 키 데이터
weight = [62, 67, 55, 74, 90, 43, 64] # 몸무게 데이터
```

```python
plt.scatter(height,weight)
plt.xlabel('height')
plt.ylabel('weight')
plt.title('Height & Weight')
plt.grid(True)
plt.show()
```

```python
# 산점도를 그릴 때 옵션인 마커 크기(s)와 컬러(c)를 동일하게 지정하는 예
plt.scatter(height,weight, s=100, c='r')
plt.xlabel('height')
plt.ylabel('weight')
plt.title('Height & Weight')
plt.grid(True)
plt.show()
```

```python
# 데이터마다 마커의 크기와 컬러를 다르게 지정하는 예
size = np.arange(1,len(height) +1) * 50
colors = ['r', 'g', 'b', 'c', 'm', 'y', 'k']
plt.scatter(height,weight, s=size, c=colors)
plt.xlabel('height')
plt.ylabel('weight')
plt.title('Height & Weight')
plt.grid(True)
plt.show()
```

```python
# 산점도를 이용하여 우리나라 주요 도시의 인구 밀도를 시각화한 예
import matplotlib
```

```python
matplotlib.rcParams['font.family'] = 'batang'
matplotlib.rcParams['axes.unicode_minus'] = False
```

```python
# 주요 도시의 이름, 위도, 경도, 인구 밀도를 순서에 맞게 리스트 데이터로 입력.
city = ['서울', '인천', '대전', '대구', '울산', '부산', '광주']

# 위도(latitude)와 경도(longitude)
lat  = [37.56, 37.45, 36.35, 35.87, 35.53, 35.18, 35.16]
lon = [126.97, 126.70, 127.38, 128.60, 129.31, 129.07, 126.85]

# 인구 밀도(명/km^2): 2017년 통계청 자료
pop_den = [16154, 2751, 2839, 2790, 1099, 4454, 2995]
```

```python
# 마커의 크기를 인구 밀도와 비례하도록 설정
soze = np.array(pop_den) * 0.2 # 마커의 크기 지정

# 도시별로 마크의 컬러를 다르게 지정
colors = ['r', 'g', 'b', 'c', 'm', 'y', 'k'] # 마커의 컬러를 지정
```

```python
plt.scatter(lon, lat, s=soze, c=colors, alpha=0.5)
plt.ylabel('latitude')
plt.xlabel('longitude')
plt.title('Population Density')
plt.grid(True)

# 마커가 위치된 곳에 도시의 이름을 표시
for x, y, name in zip(lon, lat, city):
    plt.text(x, y, name)
plt.show()
```
