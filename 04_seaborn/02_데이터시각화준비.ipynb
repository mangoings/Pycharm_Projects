```python
# 1. 라이브러리 및 데이터 읽어오기

# 2020년 건강검진 일부 데이터 'health_screenings_2020_1000ea.xlsx' 엑셀 파일을 읽어와 동적인 시각화를 표현
```

```python
import pandas as pd
```

```python
df = pd.read_excel('./input/health_screenings_2020_1000ea.xlsx')
df.head()
```

```python
df.info()
```

```python
# 데이터 전처리

# 성별, 음주여부, 흡연산태에 대하여 숫자로 저장되어 있는 정보를 데이터 분석의 가독성을 높이기 위해
# Male, Female, Non-drinking, Drinking, Non-smoking, Smoking의 문자열로 변경하여 데이터 전처리
```

```python
df1 = df[['gender', 'height', 'weight', 'waist', 'drinking', 'smoking']]
df1.head()
```

```python
df1.info()
```

```python
# 성별 데이터를 Male, Female로 변경
df1.value_counts('gender')
```

```python
df1.loc[:, 'gender'] = df1['gender'].astype('object')
df1.loc[df1['gender'] == 1, ['gender']] = 'Male'
df1.loc[df1['gender'] == 2, ['gender']] = 'Female'
```

```python
df1.value_counts('gender')
```

```python
df1.head()
```

```python
# 음주 여부 0(비음주)는 Non-drinking, 1(음주)는 Drinking으로 변경
df1.value_counts('drinking')
```

```python
df1.loc[df1['drinking'] == 0, ['drinking']] = 'Non-drinking'
df1.loc[df1['drinking'] == 1, ['drinking']] = 'Drinking'
```

```python
df1.value_counts('drinking')
```

```python
# 흡연상태 1(비흡연)과 2(흡연 끊음)은 Non-smoking, 3(흡연)은 Smoking으로 변경
df1.value_counts('smoking')
```

```python
df1.loc[df1['smoking'] == 1, ['smoking']] = 'Non-smoking'
df1.loc[df1['smoking'] == 2, ['smoking']] = 'Non-smoking'
df1.loc[df1['smoking'] == 3, ['smoking']] = 'Smoking'
```

```python
df1.value_counts('smoking')
```

```python
df1.head()
```

```python
df1.info()
```

```python
# 데이터를 저장

df1.to_pickle('./output/health_screenings_2020_1000ea.pkl')
```
