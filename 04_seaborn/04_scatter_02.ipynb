```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.rcParams.update({'figure.figsize': [8, 6]})  # 그림 크기, 기본값 [6, 4]
```

```python
df = pd.read_pickle('./output/health_screenings_2020_1000ea.pickle')
df.head()
```

```python
sns.scatterplot(data=df, x='weight', y='waist')
plt.show()
```

```python
sns.scatterplot(data=df, x='weight', y='waist', hue='gender')
plt.show()
```

```python
sns.scatterplot(data=df, x='weight', y='waist', hue='gender')
plt.show()
```

```python
import matplotlib
matplotlib.rcParams['font.family'] = 'AppleGothic'
matplotlib.rcParams['axes.unicode_minus'] = False
```

```python
sns.scatterplot(data=df, x='weight', y='waist', hue='gender')
plt.xlabel('몸무게')
plt.ylabel('허리둘레')
plt.show()
```

```python
sns.scatterplot(data=df, x='weight', y='waist', hue='gender')
plt.xlabel('몸무게')
plt.ylabel('허리둘레')
plt.show()
```

```python
df_sample = df[['weight', 'waist', 'gender']]
df_sample.head()
```

```python
# 열의 이름을 한글로
df_sample.columns = ['몸무게', '허리둘레', '성별']
df_sample.head()
```

```python
# 성별 컬럼의 값이 'Male'이면 '남성'으로 변경
df_sample.loc[:, '성별'] = df_sample['성별'].replace({'Male': '남성', 'Female': '여성'})
df_sample.head()
```

```python
sns.scatterplot(data=df_sample, x='몸무게', y='허리둘레', hue='성별')
plt.show()
```
