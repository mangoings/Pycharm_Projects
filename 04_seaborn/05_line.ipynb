# 05. 선 그래프

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

### 1. 시계열 그래프 만들기

```python
# economics 데이터 불러오기
df = pd.read_csv('./input/economics.csv')
df.head()
```

```python
df.info()
```

```python
# x축에는 시간을 나타낸 date, y축에는 실업자 수를 나타낸 unemploy를 지정.
sns.lineplot(data=df, x='date', y='unemploy')
plt.show()
```

#### 1) x축에 연도 표시하기

```python
# date 타입을 날짜 시간 타입으로 변경해서 date2 변수 추가
df['date2'] = pd.to_datetime(df['date'])
```

```python
# 변수 타입 확인
df.info()
```

```python
# 변수의 타입을 날짜 시간 타입으로 바꾸더라도 값이 달라지지 않음.
df[['date', 'date2']].head()
```

```python
# 연 추출
df['date2'].dt.year.head()
```

```python
# 월 추출
df['date2'].dt.month.head()
```

```python
# 일 추출
df['date2'].dt.day.head()
```

```python
# 연도 변수 추가
df['year'] = df['date2'].dt.year
df.head()
```

```python
df.info()
```

```python
# x축에 연도 표시
sns.lineplot(data=df, x='year', y='unemploy')
plt.show()
```

```python
# 신뢰구간 제거
sns.lineplot(data=df, x='year', y='unemploy', errorbar=None)
plt.show()
```

```python
sns.lineplot(data=df, x='date2', y='unemploy')
plt.show()
```
