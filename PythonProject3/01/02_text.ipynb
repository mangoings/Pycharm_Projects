```python
# streamlit으로 제작한 페이지에 텍스트를 표시하는 방법을 알아봄
# streamlit 웹 데시보드에서 나타낼 수 있는 텍스트들은 대표적으로 제목 title, 헤더 header, 서브헤더 subheader, 코드블럭 code block, 마크다운 markdown, 텍스트 text등이 있음
#
# 제목, 헤더, 서브헤더는 각각 대제목, 중제목, 소제목과 같이 특정 구조의 위계에 따른 제목을 붙일 때 사용
# 제목에서 서브헤더로 갈 수록 보다 세부적인 항목의 제목을 지정할 때 사용함.
```

```python
import streamlit as st

st.title('This is title')
st.header('This is header')
st.subheader('This is subheader')
```

```python
# 제목에서 헤더로 갈수록 나타나는 글씨가 작아지는 것을 확인
#
# 다음으로는 본문을 작성하는데 사용할 수 있는 마크다운
# 마크다운은 텍스트 기반의 마크업 언어로, 텍스트와 특수문자를 이용하여 텍스트 형식을 다양하게 튜닝할 수 있음
```

```python
st.markdown(
    '''
    This is main text
    This is how to change the color of text :red[Red,] :blue[Blue,] :green[Green,]
    This is **Bold** and *Italic* text
    '''
)
```
