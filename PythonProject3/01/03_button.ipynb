```python
# button 위젯은 클릭을 통해 특정 기능을 실행하거나 활성화할 수 있는 버튼
```

```python
import streamlit as st
from streamlit.file_util import streamlit_read


# button
def button_write():
    st.write('button activated')

st.button('Reset', type='primary')
st.button('activate', on_click=button_write)
```

```python
# def은 무조건 앞에 해줘야함
```

```python
# 'button activated'라는 텍스트를 나타내 주는 button_write() 함수를 정의
# 이후 st.button()을 통해 버튼을 생성
# type인자는 버튼의 형태를 조절, primary가 전달되면 빨간색 버튼을 생성. 기본값은 secondary
# on_click 인자에는 클릭 시에 동작할 함수이름을 전달
```

```python
# 'activate'가 적힌 흰색 버튼을 클릭하면 button_write() 함수가 실행되면서 'button activated' 텍스트가 나타남
# 'Reset'이 적힌 빨간색 버튼을 클릭하면 텍스트가 사라짐
# streamlit에서는 사용자가 특정 동작을 하면 (여기서는 Reset 버튼의 클릭) 해당 동작의 결과를 나타내기 위해
# 전체 스크립트가 재시작, 새로고침과 동일한 동작(이전 동작을 저장하지 않음)
# streamlit에서는 이를 세션이라고 하는데, 'Reset' 버튼에서는 클릭시 작동할 동작을 지정하지 않았기 때문에 단순히 세션만 초기화
```

```python
import streamlit as st

# 클릭 시 if문을 실행
st.button('Reset', type='primary')
if st.button('activate'):
    st.write('button activated')
```

```python
# button() 매서드는 클릭 시 True를 반환
# 동일한 결과를 얻기 위해서 위와 같이 코드를 작성할 수도 있음
```
