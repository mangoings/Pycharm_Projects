```python
# checkbox 위젯은 네모 상자를 클릭하여 체크표시를 남기고 (반대로도 가능), 해당 동작에 따른 기능들을 구현할 수 있는 위젯
# checkbox 위젯이 button 위젯과 다른 점은 사용자가 해당 위젯을 클릭하였다는 이력이 체크표시를 통해 나타남
#
# 아래 코드는 checkbox 위젯을 웹 대시보드에 표현하ㅏ고, 클릭 시 간단한 텍스트를 출력하는 기능을 구현
```

```python
import streamlit as st
from streamlit import toggle

activate = st.checkbox('I agree')

if activate:
    st.write('Great!')
```

```python
# 클릭되면 True로 반환. 해당 변수를 if문을 사용하여 참일때 'Great!'라는 텍스트를 표현하게끔 코딩
# 초기 상태는 버튼이 클릭되지 않음. 버튼을 클릭하여 checkbox를 활성화시키게 되면 빨간색으로 체크표시가 나타나게 되며,
# if문이 활성화되어 'Great!'라는 텍스트가 표현
```

```python
# 다음으로 알아볼 위젯은 toggle
# 간단한 클릭 조작으로 on / off를 쉽게 변경할 수 있는 스위치와 같은 기능
# 해당 함수는 toggle 버튼이 켜지게 되면 True를, 꺼지면 False를 반환
```

```python
toggle = st.toggle('Turn on the switch!', value=True)
if toggle:
    st.write('Switch is turned on!')
else:
    st.write('Switch is turned off!')
```

```python
# Toggle 스위치를 클릭하여 스위치를 켜게 되면 스위치가 붉은 색으로 변하며 켜지게 되고, 그 결과 if문을 동작.
# 초기화 될 때 스위치의 위치를 'value'인자를 통해 지정할 수 있음.
# 만약 'value'인자에 True를 입력하여 toggle 스위치를 생성하게 된다면 초기화 되었을 때 스위치가 켜진 상태가 됨.
```
