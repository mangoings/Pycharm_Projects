```python
# selectbox는 주어진 여러 개의 값들 중 하나를 선택할 수 있는 위젯
# 실제로 여러 웹 페이지에서 회원가입 등의 절차에서 여러 선택지를 주고 하나를 선택하게끔 할 때 많이 사용
# 비슷한 기능을 하는 위젯으로 radio 위젯이 있지만, selectbox는 선택된 항목만 보여주고 선택되지 않은 나머지 선택지는 모두 숨길 수 있어
# 웹 대시보드를 보다 깔끔하게 나타낼 수 있음
```

```python
# 해당 위젯에 대한 설명을 표시할 수 있는 'abel'인자와,
# 선택지를 전달하는 options 인자가 있음
```

```python
import streamlit as st
from streamlit import multiselect

option = st.selectbox(
    label='your selection is',
    options=['Car', 'Airplane', 'Train', 'Ship']
)
st.write(f'you selected {option}')
```

```python
# 'options' 인자에 전달된 값들 중 첫번째 값이 기본값으로 지정
# 이 때 'index' 인자에 None을 전달하게 되면 selelctbox가 아무것도 선택되지 않은 채로 초기화
# 'placeholder' 인자에 아무것도 선택되지 않았을 때 나타낼 텍스트를 문자열 형식으로 전달할 수 있음
```

```python
option = st.selectbox(
    label='your selection is',
    options=['Car', 'Airplane', 'Train', 'Ship'],
    index=None,
    placeholder='Select transportation'
)
st.write(f'you selected {option}')
```

```python
# selectbox 위젯이 아무것도 선택되지 않은 채 초기화되어 결과 테스트에 'you selected: None'이 나타난 것을 확인
# 'select transportation' 이라는 placeholder가 정상적으로 나타나는 것을 확인
#
# selectbox와 비슷한 기능을 하는 위젯으로 radio가 있음
# selectbox 위젯이 드립다운 방식으로 위젯을 클릭했을 때 선택 옵션들이 나타나는 방식이라면,
# radio는 초기화시 모든 선택지가 한 눈에 들어오고, 그 중 하나를 선택할 수 있다는 차이점이 있음
# 간단히 생성할 수 있고, 함수의 반환값으로 선택된 옵션을 반환
```

```python
option = st.radio(
    'What is your favourite movie genre',
    ["Comedy", "Drama", "Documentary"],
    captions=['Laugh out loud', 'Get the popcorn', 'Never stop learning']
)

if option:
    st.text(f'You selected {option}')
```

```python
# selectbox가 단일 항목에 대해서만 선택을 할 수 있는 반면,
# multiselect는 아무것도 선택되지 않은 경우를 포함하여, 한 개 이상의 항목을 선택할 수 있음
# selectbox 위젯과 사용 방법이 매우 유사
```

```python
option = st.multiselect(
    label='your selection is',
    options=['Car', 'Airplane', 'Train', 'Ship'],
    placeholder='Select transportation'
)
st.write(f'you selected {option}')
```

```python
# 초기화된 형태는 selectbox와 동일하지만, multiselect 위젯의 항목을 선택하면 선택된 항목들이 붉은색 박스로 나타나게 되고,
# 계속해서 다른 항목들을 추가로 선택할 수 있다.
# 이렇게 선택된 항목들은 박스 옆의 X 표시를 클릭하여 선택해제 할 수도 있음.
```
