```python
# slider는 수치형, 날짜 혹은 시간 값에 대해 그 값 자체거나, 값의 범위를 선택할 수 있게끔 하는 위젯
#
# silder라는 함수를 통해 사용할 수 있음
# 첫 번째 인자로 slider 위에 표시될 레이블을 설정할 수 있고, slider를 통해 선택할 수 있는 값의 최소, 최대, 단위를 각각 수치형 값으로 전달
# slider 위젯을 통해 선택된 값을 받아 올 변수를 함수의 반환값으로 지정
```

```python
import streamlit as st

score = st.slider('Your score is...', 0, 100, 1)
st.text(f'Score: {score}')
```

```python
# 초기 생성된 모습은 값이 0으로 세팅되어 있고, 아래쪽 텍스트도 0을 나타내고 있음
# slider를 우측으로 끌어서 값을 82에 맞추게 되면 아래 텍스트도 자동으로 82로 출력되도록 설정된 것을 확인
#
# 다음으로 특정 값의 범위를 나타낼 수 있는 slider를 생성
# slider는 수치형 값만 아니라 날짜 및 시간 값을 다룰 수도 있음
```

```python
# 이번에는 datetime 라이브러리를 이용하여 날짜와 시간 값의 범위를 slider 위젯을 통해 지정해 봄
# 단위를 나타내는 'value'인자에 길이가 2인 튜플을 전달하는 것과, 지정된 범위의 양 끝값을 반환하기 때문에 두 개의 변수로 반환값을 받을 수 있다는 차이점
#
# 아래 코드는 0시부터 23시까지의 범위 중 특정 범위를 지정할 수 있는 slider 위젯을 생성
# 또한 시간의 범위를 지정하기 위해 'value'인자에 두 개의 값을 튜플 형식으로 묶어서 전달
# 이렇게 전달하면 범위를 지정하는 slider가 생성되고, 전달된 값이 slider의 초기값으로 설정
```

```python
import streamlit as st

from datetime import time

start_time, end_time = st.slider(
    'Working time is...',
    min_value=time(0), max_value=time(23),
    value=(time(8), time(18)),
    format='HH:mm'
)
st.text(f'Working time: {start_time}, {end_time}')
```
