```python
# file_uploader 함수를 이용하여 사용자의 로컬 PC에 있는 다양한 형식의 파일들을 웹 대시보드에 업로드 할 수 있음
# csv파일 외에도 이미지 파일을 불러와 이미지 처리 및 인공지능 모델에 활용할 수 있음
#
# * file_upload 함수의 주요 인자
# 1) type : 업로드 가능한 파일의 확장자를 지정할 수 있음
# 해당 인자에 문자열이나 리스트 형식으로 확장자를 지정해 주면 해당 확장자를 가지는 파일만 업로드할 수 있도록 설정
# 2) accept_multiple_files : 한번에 여러 개의 파일을 업로드 할 수 있도록 설정할 수 있음.
# 해당 인자에 True를 전달하게 되면 동시에 여러개의 파일을 업로드할 수 있고,
# 업로드 된 파일들은 리스트 형태로 변환
# 3) on_change : 해당 인자에 특정 함수를 전달하게 되면 file_upload 위젯의 값이 바뀔 때마다 전달된 함수를 호출
```

```python
import pandas as pd
import streamlit as st

# file uploader
file = st.file_uploader(
    'Choose a file',
    type=['csv'],
    accept_multiple_files=False
)

if file is not None:
    df = pd.read_csv(file)
    st.write(df)
```

```python
# 인자에 type에 csv를, accept_multiple_files에 False를 전달
# 해당 함수의 반환값을 file 변수로 받아오고, 해당 변수가 None이 아닐 때
# pandas의 read_csv 함수를 이용하여 가져온 csv 파일을 df 변수에 저장하고
# Streamlit write 함수를 이용하여 표현
```
