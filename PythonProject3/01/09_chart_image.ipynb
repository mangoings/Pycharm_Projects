```python
# 1. 차트
#
# streamlit 라이브러리는 내장 함수를 통해 scatterplot, lineplot, barplot 등과 같은 기본적인 그래프를 그리는 기능을 제공
# 하지만 복잡한 그래프를 그리기 위해서는 matplotlib, seaborn 같은 라이브러리가 필요한데,
# streamlit은 시각화 전용 라이브러리로 그린 그래프를 웹 대시보드에 나타낼 수 있도록 하는 기능을 제공
#
# seaborn 라이브러리의 load_dataset 메서드를 이용하여 tips 데이터셋을 불러와서 x축을 total_bill로 하여 히스토그램을 그림
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import streamlit as st
from matplotlib import matplotlib_fname

df = sns.load_dataset('tips') # 기본적으로 제공해주는 기본 데이터(tips)

fig, ax = plt.subplots()
sns.histplot(df, x='total_bill', ax=ax, hue='time')

# streamlit 대시보드에 표현
st.pyplot(fig)
```
