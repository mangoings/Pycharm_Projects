```python
# 1. 사이드바
#
# 사이드바는 페이지 레이아웃에서 주로 본문의 왼쪽에 메뉴 등과 같이 페이지의 부가적인 요소를 담고 있는,
# 본문과는 분리된 공간을 뜻함
# 웹 포털 사이트의 블로그에서 비슷한 구조를 쉽게 찾아볼 수 있음
#
# Streamlit에서 사이드바는 with 구문을 이용하여 생성할 수 있음
# with 구문안에 사이드바에 표시할 텍스트나 위젯 등 다양한 요소를 포함시키면 사이드바와 본문을 구분할 수 있음
```

```python
import streamlit as st
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from PIL import Image

# slider
st.title('This is main page')

with st.sidebar:
    st.title('This is sidebar')
    side_option = st.multiselect(
        label='your selection is',
        options=['Car', 'Airplane', 'Train', 'Ship', 'Bicycle'],
        placeholder='select transportation'
    )
```

```python
# 완성된 웹 대시보드의 왼편에 회색 바탕의 사이드바가 생성되었고,
# 사이드바 내부에 with 구문 내에 넣어주었던 multiselect 위젯이 생성
# 사이드바 with 구문 밖에 위치한 타이틀 텍스트는 본문에 정상적으로 표시
#
# 사이드바와 본문의 경계를 마우스로 드래그하여 사이드바의 크기를 조절할 수도 있고,
# 사이드바 우측 상단의 X 표시를 클릭하면 사이드바를 숨김 처리할 수도 있음
```

```python
# 위젯이나 텍스트 등 각종 Streamlit 요소들은 위에서 아래 방향으로 나타남
# 만약 특정 요소들을 위에서 아래 방향이 아니라 왼쪽에서 오른쪽으로 수평 방향으로 나타내고자 한다면 열 요소를 사용할 수 있음
```

```python
# column 이해
img2 = Image.open('./input/img2.jpg')
img3 = Image.open('./input/img3.jpg')

st.header('Lemonade')
st.image(img2, width=300, caption='Image from Unsplash')

st.header('Cocktail')
st.image(img3, width=300, caption='Image from Unsplash')
```

```python
# PIL 라이브러리를 이용하여 두 개의 이미지를 가져와서 일반적인 위에서 아래 순서로 표시하는 코드
# 세로로 긴 두 개의 이미지를 위에서 아래 방향으로 표시하게 되니 여백이 커서 페이지 낭비가 심함
```

```python
# column 활용
col1, col2 = st.columns(2)

with col1:
    st.header('Lemonade')
    st.image(img2, width=300, caption='Image from Unsplash')

with col2:
    st.header('Cocktail')
    st.image(img3, width=300, caption='Image from Unsplash')
```

```python
# 두 개의 이미지를 열로 구분하여 왼쪽에서 오른쪽 수평 방향으로 나타내는 코드
# Streamlit의 column 메서드를 이용하여 나누어지는 각각의 열을 col1, col2 변수로 가져옴
# 이 때 columns 메서드에 전달되는 인자는 나누어 줄 열의 개수
# 각각의 변수를 with 구문으로 묶어주게 되면 구문 내에 묶인 요소들은 해당 열에 표시
```

```python
# 3. tab
```

```python
# 탭은 여러 종류의 구성요소들 중 특정 그룹만 표시하고, 나머지는 숨김 처리
# 엑셀의 탭 메뉴와 비슷
```

```python
# tab
tab1, tab2 = st.tabs(['Table', 'Graph'])

df = pd.read_csv('./input/medical_cost.csv')
df = df.query('region == "northwest"')

with tab1:
    st.table(df.head(5))

with tab2:
    fig, ax = plt.subplots()
    sns.scatterplot(data=df, x='bmi', y='charges', ax=ax)
    st.pyplot(fig)
```

```python
# 두개의 탭을 만든 후, medical_cost 데이터 셋을 불러와서 northwest 지역에 해당하는 데이터들만 필터링 후,
# 첫 번째 탭에는 첫 5개의 행을 표시하는 데이터 테이블을,
# 나머지 탭에는 필터링한 데이터셋에서 x축을 bmi 변수로, y축을 charge 변수로 하는 scatterplot을 표현
# 각각의 탭을 생성하기 위해 Streamlit의 tabs 메서드를 이용
#
# 탭 구성요소는 동시에 표시하기에는 페이지의 낭비가 큰 구성요소들을 각각의 탭으로 나눠서 표시할 수 있음
```

```python
# 4. expander
```

```python
# expander는 특정 구성 요소들을 묶어서 숨김 및 펼치기 처리를 하여 보다 효율적인 대시보드 구성을 하기 위한 요소
# with 구문을 이용해서 사용할 수 있고, with 구문 안에 코딩된 요소들에 한해 숨기기 및 펼치기 처리를 할 수 있음
```

```python
# expander
df = pd.read_csv('./input/medical_cost.csv')
df = df.query('region == "northwest"')

fig, ax = plt.subplots()
sns.scatterplot(data=df, x='bmi', y='charges', ax=ax)
st.pyplot(fig)

with st.expander("See datatable"):
    st.table(df.head(5))
```

```python
# expander 메소드의 인자로 전달된 문자열 "See datatable"은 expander의 레이블로 표시
# expander를 클릭하여 펼치면 with 구문 안에 코딩하였던 데이터테이블이 나타남
# 모든 경우에 확인할 필요 없이 필요에 따라 펼쳐서 확인할 수 있는 요소들을 숨겨 대시보드를 보다 깔끔하게 관리할 수 있다는 장점
```
