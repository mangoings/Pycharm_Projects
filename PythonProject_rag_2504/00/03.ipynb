```python
from langchain_openai import ChatOpenAI

# 객체 생성
llm = ChatOpenAI(
    temperature=0.1, # 창의성 (0.0 ~ 2.0) ０으로 갈수록 정답 중심, 숫자가 커질수록 창의적인 답변
    max_tokens=2048, # 최대 토큰수
    model_name="gpt-3.5-turbo" # 모델명
)
```

```python
# 질의내용
question: str = '''질문에 대하여 100자 내외로 답변해 주세요
대한민국의 수도는 어디인가요?''' # : str 은 없어도 되지만 사용자가 코드를 이해하기 편하기 위해 사용 (생략가능)
```

```python
llm.invoke(question) # invoke: 작동시키다. 즉, 내가 만든 ChatOpenAI를 작동(답변하도록)시키겠다.
# 답변을 받기위해선 질문이 (괄호)안에 들어가면 됨.
```

```python
# 간략한 버전
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    temperature=0.1, max_tokens=2048, model_name="gpt-3.5-turbo"
)

response = llm.invoke(question)
print(response)
```
