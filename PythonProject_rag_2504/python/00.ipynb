```python
def sum_list(nums):
    if not isinstance(nums, (list, tuple)):
        raise TypeError('not implemented')
    if not all(isinstance(n, (int, float)) for n in nums): # 리스트나 튜플 둘 중 하나면 return
        raise ValueError("모든 요소는 정수 또는 실수여야 합니다.")
    return sum(nums)
```
